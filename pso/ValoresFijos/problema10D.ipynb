```python
import numpy as np
```

# la funcion es 
$\sum_{i=1}^{n=10} x^2-10\cos(2π x_i)+10$

```python
def f(X):
    return np.sum(X**2 - 10 * np.cos(2 * np.pi * X) + 10, axis=1)
```

```python
# Parametros
dimensiones = 10
iteraciones = 100
particulas = 100

# Ya definidos
c1 = 1.0
c2 = 1.0

Wmax = 0.9
Wmin = 0.4
```

```python
#Inicializacion de enjambre
posiciones = np.random.uniform(Wmin, Wmax, (particulas, dimensiones))
velocidades = np.random.uniform(-1,1,(particulas, dimensiones))
```

```python
# pBest
pMejoresPosiciones = posiciones.copy()
pMejoresValores = f(posiciones)
```

```python
# gbest
indiceMejorGlobal = np.argmin(pMejoresValores)
gMejorPosicion = pMejoresPosiciones[indiceMejorGlobal].copy()
gMejorValor = pMejoresValores[indiceMejorGlobal]
```

```python
# Bucle pdel PSO
print("Valor inicial: ", gMejorValor)

for i in range(iteraciones):

    # valor de w dinamico
    w = Wmax - (Wmax - Wmin) * i / iteraciones

    # numeros aleatorios para la estocasidad el algoritmo
    r1 = np.random.rand(particulas, dimensiones)
    r2 = np.random.rand(particulas, dimensiones)

    velocidades = (w * velocidades + 
                   c1 * r1 * (pMejoresPosiciones - posiciones) +
                   c2 * r2 * (gMejorPosicion - posiciones))

    # actualizar posiciones
    posiciones = posiciones + velocidades

    # evaluacion de la aptitud (firness)
    valores_actuales = f(posiciones)

    # Actualizacion de pBest en caso de que sea mejor
    mejores_mask = valores_actuales < pMejoresValores
    pMejoresValores[mejores_mask] = valores_actuales[mejores_mask]
    pMejoresPosiciones[mejores_mask] = posiciones[mejores_mask]

    # Actualizacion para el gBest, el mejor global
    indiceMejorGlobal = np.argmin(pMejoresValores)
    if pMejoresValores[indiceMejorGlobal] < gMejorValor:
        gMejorValor = pMejoresValores[indiceMejorGlobal]
        gMejorPosicion = pMejoresPosiciones[indiceMejorGlobal].copy()
```

```python
print("mejor posicion:", gMejorPosicion)
print("Valor minimo: ", gMejorValor)
```
