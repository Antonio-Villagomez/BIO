```python
import numpy as np

# 1. Definir la función objetivo a minimizar (Función Esfera)
def esfera(x):
    return np.sum(x**2, axis=1)

# 2. Configuración de hiperparámetros del PSO
num_particulas = 30
dimensiones = 2
iteraciones = 50

# Límites del espacio de búsqueda (ej. entre -10 y 10)
limite_inferior = -10.0
limite_superior = 10.0

w = 0.5   # Inercia (peso de la velocidad anterior)
c1 = 1.5  # Coeficiente cognitivo (atracción hacia el pBest)
c2 = 1.5  # Coeficiente social (atracción hacia el gBest)

# 3. Inicialización del Enjambre
# Posiciones aleatorias dentro de los límites
posiciones = np.random.uniform(limite_inferior, limite_superior, (num_particulas, dimensiones))
# Velocidades iniciales cercanas a cero
velocidades = np.random.uniform(-1, 1, (num_particulas, dimensiones))

# Inicializar los mejores personales (pBest) con las posiciones actuales
p_best_posiciones = posiciones.copy()
p_best_valores = esfera(posiciones)

# Inicializar el mejor global (gBest)
indice_mejor_global = np.argmin(p_best_valores)
g_best_posicion = p_best_posiciones[indice_mejor_global].copy()
g_best_valor = p_best_valores[indice_mejor_global]

# 4. Bucle principal del PSO
print(f"Valor inicial global óptimo: {g_best_valor:.6f}")

for it in range(iteraciones):
    # Generar números aleatorios para la estocasidad del algoritmo
    r1 = np.random.rand(num_particulas, dimensiones)
    r2 = np.random.rand(num_particulas, dimensiones)
    
    # Actualizar velocidad
    velocidades = (w * velocidades + 
                   c1 * r1 * (p_best_posiciones - posiciones) + 
                   c2 * r2 * (g_best_posicion - posiciones))
    
    # Actualizar posición
    posiciones = posiciones + velocidades
    
    # Evaluar la nueva aptitud (fitness)
    valores_actuales = esfera(posiciones)
    
    # Actualizar el mejor personal (pBest) si la posición actual es mejor
    mejores_mask = valores_actuales < p_best_valores
    p_best_valores[mejores_mask] = valores_actuales[mejores_mask]
    p_best_posiciones[mejores_mask] = posiciones[mejores_mask]
    
    # Actualizar el mejor global (gBest)
    indice_minimo_actual = np.argmin(p_best_valores)
    if p_best_valores[indice_minimo_actual] < g_best_valor:
        g_best_valor = p_best_valores[indice_minimo_actual]
        g_best_posicion = p_best_posiciones[indice_minimo_actual].copy()

# 5. Resultados finales
print("\n--- Optimización Finalizada ---")
print(f"Mejor posición encontrada (x, y): {g_best_posicion}")
print(f"Valor mínimo de la función: {g_best_valor:.10f}")
```
